```python
# Find topper in physics ,maths n chemistry sub

marks = { 1: [78, 97, 86], 2: [67, 89, 90], 3: [78, 90, 87] }

subjects = ["Physics", "Maths", "Chemistry"]

for subject_index in range(3):
    topper = None
    highest_mark = -1

    for student, student_marks in marks.items():
        if student_marks[subject_index] > highest_mark:
            highest_mark = student_marks[subject_index]
            topper = student

    if topper is not None:
        print(f"{subjects[subject_index]} topper: Student {topper} with {highest_mark} marks")
    else:
        print(f"No marks found for {subjects[subject_index]}")
```
